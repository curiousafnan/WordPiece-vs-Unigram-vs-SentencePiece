```python
import regex as re
```

```python
corpus = "I love Pakistan and I am a pakistani. We are very proud nation. As pakistani guy, I love playing and watching cricket. Hurray!!"
pattern = re.compile(
    r"""'(?:[sdmt]|ll|ve|re)|\p{L}+|\p{N}+|[^\s\p{L}\p{N}]"""
)
start = "##"
corpus = pattern.findall(corpus)
# corpus
# corpus = [list(chunk) for chunk in corpus]
# print(corpus)
text = []
for word in corpus:
    temp = []
    for idx, char in enumerate(word):
        if idx == 0:
            temp.append(char)
            continue
        temp.append(start+char)
    text.append(temp)
print(text)
```
